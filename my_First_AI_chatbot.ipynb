```python
from google.colab import userdata
Open_ai_api_key = userdata.get("Open_ai_api_key")
print("api key is configured successfully, yayyyyyy")
```

```python
from google import genai
client = genai.Client(api_key = Open_ai_api_key)
print("gemini is connnected to python")
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Explain what a Large Language Model is in simple words."
)

print(response.text)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="I don't understand llm, explain those like I am begineer"
)

print(response.text)
```

```python
while True:
  user_input = input("you:")
  if user_input.lower() == "exit":
    print("Bot: Good bye")
    break
  response = client.models.generate_content(
      model = "gemini-3.5-flash-lite",
      contents = user_input
)
  print("Bot:",response.text)
```

```python
chat = client.chats.create(
    model = "gemini-3.5-flash-lite"
)
print("Welcome to bindu chatbot")
print("type exit to stop \n")
while True:
  user_input = input("You:")
  if user_input.lower() == "exit":
    print("Bot:goodbye")
    break
  try:
    response = chat.send_message(user_input)
    print(response.text)
  except Exception as e:
    print("something went wrong:",e)
```
